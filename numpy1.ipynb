```python
import numpy as np
```

```python
list=[1,2,3,4,5]
print(list)
type(list)
```

```python
print('1D Array')
a=np.array([1,2,3,4,5])
print(a)
type(a)
```

```python
print('2D Array')
b=np.array([[1,2,3,4,5],[9,8,7,6,5]])
print(b)
type(b)
```

```python
print('3D Array')
c=np.array([[[1,2,3,4,5],[9,8,7,6,5],[2,5,6,7,7]]])
print(c)
type(c)
```

```python
print(a.size)
print(b.size)
print(c.size)
```

```python
print(a.shape)
print(b.shape)
print(c.shape)
```
